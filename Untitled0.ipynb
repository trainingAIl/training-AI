```python
import torch
import torch.nn as nn

class SmallAI(nn.Module):
    def __init__(self, vocab_size, embed_dim=32, hidden_dim=64):
        super().__init__()
        self.embedding = nn.Embedding(vocab_size, embed_dim)
        self.lstm = nn.LSTM(embed_dim, hidden_dim, batch_first=True)
        self.output = nn.Linear(hidden_dim, vocab_size)

    def forward(self, x):
        x = self.embedding(x)
        x, _ = self.lstm(x)
        return self.output(x)

print("AI 模型已準備好！")

```

```python
import csv
import torch
import torch.nn as nn
from torch.utils.data import DataLoader, TensorDataset

# 讀取你剛建立的訓練資料
with open("qa_dataset.csv", newline="", encoding="utf-8") as f:
    rows = list(csv.DictReader(f))

questions = [row["question"] for row in rows]
answers = [row["answer"] for row in rows]

# 建立詞彙表：把文字變成數字
words = set()
for question in questions:
    words.update(question.lower().replace("?", "").split())

word_to_id = {"<pad>": 0, "<unk>": 1}
for word in sorted(words):
    word_to_id[word] = len(word_to_id)

# 把每個問題變成固定長度的數字序列
def encode(question, max_len=10):
    ids = [word_to_id.get(word, 1) for word in question.lower().replace("?", "").split()]
    ids = ids[:max_len]
    ids += [0] * (max_len - len(ids))
    return ids

# 每個答案給一個編號，例如 0 = 姓名、1 = 生日、2 = 身高
answer_to_id = {}
id_to_answer = {}

for answer in answers:
    if answer not in answer_to_id:
        answer_to_id[answer] = len(answer_to_id)
        id_to_answer[len(id_to_answer)] = answer

X = torch.tensor([encode(q) for q in questions], dtype=torch.long)
y = torch.tensor([answer_to_id[a] for a in answers], dtype=torch.long)

dataset = TensorDataset(X, y)
loader = DataLoader(dataset, batch_size=4, shuffle=True)

# 建立模型
model = SmallAI(len(word_to_id), embed_dim=32, hidden_dim=64)

loss_function = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(model.parameters(), lr=0.01)

# 開始訓練
epochs = 100

for epoch in range(1, epochs + 1):
    total_loss = 0

    for batch_x, batch_y in loader:
        optimizer.zero_grad()

        prediction = model(batch_x)
        loss = loss_function(prediction[:, -1, :], batch_y)

        loss.backward()
        optimizer.step()

        total_loss += loss.item()

    if epoch == 1 or epoch % 10 == 0:
        print(f"Epoch {epoch:3d} | Loss: {total_loss / len(loader):.4f}")

print("訓練完成！")
```

```python
def ask(question):
    model.eval()

    with torch.no_grad():
        x = torch.tensor([encode(question)], dtype=torch.long)
        prediction = model(x)
        answer_id = int(prediction[:, -1, :].argmax(dim=-1))
        answer = id_to_answer[answer_id]

    return answer

# 測試訓練時看過的問題
print(ask("What is my height?"))
print(ask("What is my shoe size?"))
print(ask("What is my favorite subject?"))

# 測試沒看過的問法
print(ask("Could you remind me how tall I am?"))
print(ask("What size shoes do I wear?"))
```

```python
import json

# 儲存模型參數
torch.save(model.state_dict(), "my_ai_model.pt")

# 儲存文字轉數字的對照表
with open("word_to_id.json", "w", encoding="utf-8") as f:
    json.dump(word_to_id, f, ensure_ascii=False, indent=2)

# 儲存答案編號對照表
with open("id_to_answer.json", "w", encoding="utf-8") as f:
    json.dump(id_to_answer, f, ensure_ascii=False, indent=2)

print("模型與資料已成功儲存！")
```

```python
# 重新建立一個空的模型
loaded_model = SmallAI(len(word_to_id), embed_dim=32, hidden_dim=64)

# 載入你剛訓練好的權重
loaded_model.load_state_dict(torch.load("my_ai_model.pt"))
loaded_model.eval()

def ask_loaded(question):
    with torch.no_grad():
        x = torch.tensor([encode(question)], dtype=torch.long)
        prediction = loaded_model(x)
        answer_id = int(prediction[:, -1, :].argmax(dim=-1))
        return id_to_answer[answer_id]

print(ask_loaded("What is my height?"))
print(ask_loaded("What is my shoe size?"))
```
